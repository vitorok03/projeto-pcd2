```python
import pandas as pd
from pathlib import Path

from soja_clima.clima import carregar_e_agregar_inmet
```

```python
RAW_DIR = Path("../../data/raw")
PROCESSED_DIR = Path("../../data/processed")
PASTA_INMET = RAW_DIR / "inmet"
```

```python
# Só o clima diário: o mapa estação -> microrregião é gerado em memória
# pelo main.ipynb (soja_clima.geo.construir_mapa_estacao_microrregiao).
df_clima_diario = carregar_e_agregar_inmet(str(PASTA_INMET))
```

```python
df_clima_diario.to_parquet(PROCESSED_DIR / "clima_diario.parquet", index=False)

print("Clima diário salvo com sucesso!")
```
